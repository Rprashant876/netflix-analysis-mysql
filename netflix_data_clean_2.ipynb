```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv("D:\\sql\\dataset\\NetFlix.csv")
data
```

```python
data.isnull().sum()
```

```python
data['director']=data['director'].fillna('unknown')
data['cast']=data['cast'].fillna('not available')
data['country']=data['country'].fillna('unknown')
data['rating']=data['rating'].fillna('not rated')
data['date_added'] = data['date_added'].fillna(pd.to_datetime(data['release_year'], format='%Y'))
```

```python
data.isnull().sum()
```

```python
data['date_added'] = pd.to_datetime(data['date_added'], errors='coerce').dt.date
```

```python
data.to_csv('netflix_cleaned.csv',index=False)
```
